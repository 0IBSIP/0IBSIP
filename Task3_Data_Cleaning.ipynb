## 4. Dataset Loading

```python
# ============================================================
# CELL 2 — LOAD DATASET
# ============================================================

import pandas as pd

# Directly load the uploaded dataset
file_name = "train.csv"

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File Name:", file_name)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

## 6. Data Quality Assessment

Before cleaning the dataset, a data-quality assessment is performed.

The assessment checks:

- Missing values
- Missing-value percentage
- Data types
- Unique values
- Duplicate records

This establishes a baseline against which the final cleaned dataset
can be compared.

```python
# ============================================================
# CELL 4 — DATA QUALITY REPORT
# ============================================================

quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Null_Count": df.isnull().sum().values,
    "Null_Percentage": (df.isnull().mean() * 100).round(2).values,
    "Unique_Values": df.nunique().values
})

print("========== DATA QUALITY REPORT ==========")

display(quality_report)

print("\n========== SUMMARY ==========")
print("Total Rows        :", len(df))
print("Total Columns     :", len(df.columns))
print("Total Null Values :", df.isnull().sum().sum())
print("Duplicate Rows    :", df.duplicated().sum())
```

## 7. Numerical Range Analysis

Numerical columns are examined to identify unusual or potentially
invalid values.

The analysis includes:

- Minimum
- Maximum
- Mean
- Median
- Standard deviation

This provides an initial understanding of the numerical distribution
before outlier detection.

```python
# ============================================================
# CELL 5 — NUMERIC RANGE REPORT
# ============================================================

numeric_cols = df.select_dtypes(include=np.number).columns

range_report = pd.DataFrame({
    "Column": numeric_cols,
    "Minimum": [df[col].min() for col in numeric_cols],
    "Maximum": [df[col].max() for col in numeric_cols],
    "Mean": [df[col].mean() for col in numeric_cols],
    "Median": [df[col].median() for col in numeric_cols],
    "Std_Dev": [df[col].std() for col in numeric_cols]
})

display(range_report)
```

## 8. Establishing the Before-Cleaning Baseline

Before modifying the dataset, important quality metrics are stored.

These metrics will later be compared with the cleaned dataset to
quantify the impact of the cleaning process.

```python
# ============================================================
# CELL 6 — BEFORE-CLEANING METRICS
# ============================================================

before_rows = len(df)
before_columns = len(df.columns)
before_nulls = df.isnull().sum().sum()
before_duplicates = df.duplicated().sum()

print("========== BEFORE CLEANING ==========")
print("Rows              :", before_rows)
print("Columns           :", before_columns)
print("Total Null Values :", before_nulls)
print("Duplicate Rows    :", before_duplicates)
```

## 9. Duplicate Detection and Removal

Duplicate records can cause observations to be counted multiple times
and may affect statistical analysis.

Exact duplicate rows are therefore identified and removed.

The number of duplicate rows removed is documented for transparency.

```python
# ============================================================
# CELL 7 — DUPLICATE REMOVAL
# ============================================================

duplicate_count = df.duplicated().sum()

print("Duplicate rows found:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", duplicate_count)
print("Rows after duplicate removal:", len(df))
```

## 10. Column Name Standardization

Consistent column names make the dataset easier to work with and reduce
errors during analysis.

The column names are:

- converted to lowercase
- stripped of unnecessary spaces
- standardized using underscores

```python
# ============================================================
# CELL 8 — STANDARDIZE COLUMN NAMES
# ============================================================

df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

print("Standardized column names:")
print(df.columns.tolist())
```

## 11. Text and Categorical Standardization

Inconsistent text formatting can result in duplicate categories.

For example:

`Male`, `male`, and `M`

may represent the same category.

Text fields are therefore stripped of unnecessary spaces and important
categorical values are standardized.

```python
# ============================================================
# CELL 9 — STANDARDIZE TEXT VALUES
# ============================================================

text_columns = [
    "name",
    "sex",
    "ticket",
    "cabin",
    "embarked"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Standardize Sex
df["sex"] = (
    df["sex"]
    .str.lower()
    .replace({
        "m": "male",
        "male": "male",
        "f": "female",
        "female": "female"
    })
)

# Standardize Embarked
df["embarked"] = df["embarked"].str.upper()

print("Text standardization completed.")
```

## 12. Missing Value Analysis

Missing values are inspected before treatment.

Different columns require different strategies depending on their
data type and meaning.

The following strategies are used:

| Column | Strategy | Reason |
|---|---|---|
| Age | Median | Numerical variable and robust to extreme values |
| Embarked | Mode | Categorical variable |
| Cabin | Unknown | Preserve observations and retain missingness information |

```python
# ============================================================
# CELL 10 — MISSING VALUE REPORT
# ============================================================

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Values": df.isnull().sum().values,
    "Missing_Percentage":
        (df.isnull().mean() * 100).round(2).values
})

missing_report = missing_report[
    missing_report["Missing_Values"] > 0
]

print("========== MISSING VALUES BEFORE TREATMENT ==========")

display(missing_report)
```

```python
# ============================================================
# CELL 11 — MISSING VALUE TREATMENT
# ============================================================

# Age → Median
age_median = df["age"].median()

df["age"] = df["age"].fillna(age_median)

# Embarked → Mode
embarked_mode = df["embarked"].mode()[0]

df["embarked"] = df["embarked"].fillna(embarked_mode)

# Cabin → Unknown
df["cabin"] = df["cabin"].fillna("Unknown")

print("Missing value treatment completed.")

print("\nRemaining missing values:")
display(df.isnull().sum())
```

## 13. Data Type Correction

Correct data types are important for reliable analysis.

`PassengerId` is an identifier, so it is treated as a string rather
than a numerical measurement.

Numerical variables are converted to appropriate numerical types and
text variables are stored as strings.

```python
# ============================================================
# CELL 12 — DATA TYPE CORRECTION
# ============================================================

# Identifier
df["passengerid"] = df["passengerid"].astype("string")

# Text columns
for col in ["name", "sex", "ticket", "cabin", "embarked"]:
    df[col] = df[col].astype("string")

# Numeric columns
numeric_columns = [
    "survived",
    "pclass",
    "age",
    "sibsp",
    "parch",
    "fare"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print("Data types corrected successfully.")

display(df.dtypes.to_frame("Final Data Type"))
```

## 14. Outlier Detection Using IQR

The Interquartile Range (IQR) method is used to identify statistical
outliers in numerical columns.

The analysis is performed on:

- Age
- SibSp
- Parch
- Fare

For each variable, Q1, Q3, IQR, lower bound, upper bound, and number
of detected outliers are calculated.

```python
# ============================================================
# CELL 13 — IQR OUTLIER DETECTION
# ============================================================

outlier_columns = [
    "age",
    "sibsp",
    "parch",
    "fare"
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

    outlier_count = outlier_mask.sum()

    outlier_results.append({
        "Column": col,
        "Q1": round(Q1, 2),
        "Q3": round(Q3, 2),
        "IQR": round(IQR, 2),
        "Lower_Bound": round(lower_bound, 2),
        "Upper_Bound": round(upper_bound, 2),
        "Outlier_Count": int(outlier_count),
        "Outlier_Percentage":
            round(outlier_count / len(df) * 100, 2)
    })

outlier_report = pd.DataFrame(outlier_results)

print("========== IQR OUTLIER REPORT ==========")

display(outlier_report)
```

## 15. Outlier Treatment Decision

The IQR method identifies statistical outliers, but an outlier is not
automatically an incorrect value.

For this dataset, detected outliers are retained because extreme values
may represent legitimate passenger characteristics.

For example, a high Fare value may represent a genuine higher-priced
ticket.

Therefore, the project does not delete statistical outliers without
evidence that they are invalid.

This approach reduces unnecessary information loss.

## 16. Logical Data Validation

After cleaning, logical validation checks are performed.

The purpose is to identify values that are impossible or inconsistent
with the expected meaning of the variables.

```python
# ============================================================
# CELL 14 — LOGICAL VALIDATION
# ============================================================

print("========== LOGICAL VALIDATION ==========")

print(
    "Negative Age values:",
    (df["age"] < 0).sum()
)

print(
    "Negative Fare values:",
    (df["fare"] < 0).sum()
)

print(
    "Invalid Pclass values:",
    (~df["pclass"].isin([1, 2, 3])).sum()
)

print(
    "Invalid Survived values:",
    (~df["survived"].isin([0, 1])).sum()
)

print(
    "Invalid Embarked values:",
    (~df["embarked"].isin(["C", "Q", "S"])).sum()
)
```

## 17. Final Missing Value Validation

After applying the selected missing-value strategies, the dataset is
checked again.

The goal is to verify that the intended missing values have been
successfully handled.

```python
# ============================================================
# CELL 15 — FINAL NULL CHECK
# ============================================================

final_null_report = pd.DataFrame({
    "Column": df.columns,
    "Null_Count": df.isnull().sum().values,
    "Null_Percentage":
        (df.isnull().mean() * 100).round(2).values
})

print("========== FINAL NULL REPORT ==========")

display(final_null_report)

print(
    "Total Remaining Null Values:",
    df.isnull().sum().sum()
)
```

```python
# ============================================================
# CELL 16 — FINAL DUPLICATE CHECK
# ============================================================

final_duplicates = df.duplicated().sum()

print("========== FINAL DUPLICATE CHECK ==========")
print("Remaining Duplicate Rows:", final_duplicates)
```
